### Partie 8 : Analyse

### Partie 9 : Gestion des valeurs manquantes 

##### 1) Afficher le nombre de valeurs manquantes par colonne 

```python
nb_val_manquantes = df.isnull().sum()
print(nb_val_manquantes)
```

***explication***: ce resultat represente le nombre de valeur manquante pour chaque colonne . si une colonne ne contient pas de valeurs manquantes ou des champs vide , alors le compteur est egal a 0 et si c'est le cas contraire il additionne le nombre de valeur manquante par colonne . comme exemple la colonne "temperature" contient 6 valeurs manquantes et pratiquement c'est la colonne qui renferme plus de valeurs manquantes .

##### 2) Afficher le nombre total des valeurs manquantes 

```python
nb_total_val_manquantes = df.isnull().sum().sum()
print(nb_total_val_manquantes)
```

##### 3) Afficher le taux de valeurs manquantes 

```python
df.shape[0]
```

```python
taux= (nb_total_val_manquantes/df.shape[0])*100
print(f"le taux de valeurs manquantes est : {taux:.2f}")
```

##### 4) Afficher les lignes contenant des valeurs manquantes 

```python
lignes_val_manquantes = df[df.isna().any(axis=1)]
print(lignes_val_manquantes)
```

##### 5) Pour la température et l’humidité, remplacer les valeurs manquantes par les moyennes 

```python
df['temperature'] = df['temperature'].fillna(df['temperature'].mean())
print(df['temperature'])
```

```python
df['humidite_relative'] = df['humidite_relative'].fillna(df['humidite_relative'].mean())
print(df['humidite_relative'])
```

##### 6) Pour la consommation, remplacer les valeurs manquantes par la médiane

```python
df['consommation']=df['consommation'].fillna(df['consommation'].median())
print(df['consommation'])
```
