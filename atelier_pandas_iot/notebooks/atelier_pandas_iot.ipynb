### Partie 1 : SERIES

```python
import pandas as pd
```

##### 1) Créer une Series contenant 4 températures

```python
temperature = pd.Series([20, 25, 32, 36],
index=["1", "2", "3", "4"])
print(temperature)
```

##### 3) Afficher la 1ère valeur 

```python
print(temperature["1"])
```

##### 4) Afficher la dernière valeur 

```python
print(temperature["4"])
```

##### 5) Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h 

##### premiere facon

```python
temp_1 = pd.Series([15, 20, 35, 40],
index= ["12h", "13h", "14h", "15h"])
print(temp_1)
```

##### Deuxieme facon

```python
temp_2 = {
    "12h":15,
    "13h":20,
    "14h":35,
    "15h":40
}
temp_2= pd.Series(temp_2)
print(temp_2)
```

##### 6) Afficher la 1ère et la dernière valeur 

```python
print(temp_2["12h"])
print(temp_2["15h"])
```
