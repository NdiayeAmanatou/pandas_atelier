##### Partie 5 – Manipulation des colonnes 

##### 1) Ajouter la colonne "temperature_fahrenheit" 

```python
df['temperature_fahrenheit']= 0
print(df)
```

##### 2) Ajouter la colonne "niveau_temperature" pour catégoriser la température en Élevée (température > 30) ou Normale". 

```python
df['niveau_temperature']= np.where(df['temperature'] > 30, 'Elevee', 'Normale')
df.head()
```

##### 3) Renommer la colonne "humidite" en "humidite_relative" 

```python
df = df.rename(columns={'humidite': 'humidite_relative'})
print(df)
```

##### 4) Supprimer la colonne "temperature_fahrenheit" 

```python
df = df.drop('temperature_fahrenheit', axis=1)
print(df)
```

### Partie 6 – Filtrage 

##### 1) Récupérer uniquement les lignes telles que température > 30 

```python
sup = df[df['temperature'] > 30]
print(sup)
```

##### 2) Récupérer uniquement les lignes telles température > 30 et humidité > 70 

```python
lignes = df[(df['temperature'] > 30) &  (df['humidite_relative'] > 70)]
print(lignes)
```
